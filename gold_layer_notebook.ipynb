# **Access data**

```python

spark.conf.set(
    f"fs.azure.account.auth.type.{storage_account}.dfs.core.windows.net",
    "OAuth"
)

spark.conf.set(
    f"fs.azure.account.oauth.provider.type.{storage_account}.dfs.core.windows.net",
    "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider"
)

spark.conf.set(
    f"fs.azure.account.oauth2.client.id.{storage_account}.dfs.core.windows.net",
    client_id
)

spark.conf.set(
    f"fs.azure.account.oauth2.client.secret.{storage_account}.dfs.core.windows.net",
    client_secret
)

spark.conf.set(
    f"fs.azure.account.oauth2.client.endpoint.{storage_account}.dfs.core.windows.net",
    f"https://login.microsoftonline.com/{tenant_id}/oauth2/token"
)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
trip_zone = spark.read.format("parquet").option("inferSchema", "true").option("header", "true").load(f"abfss://silver@nyctaxidataforproject.dfs.core.windows.net/trip_zone/")
```

```python
trip_zone.display()
```

##  **Create Database**

```python
%sql
CREATE DATABASE gold
```

```python
trip_zone.write.format("delta").mode('append').option("path","abfss://gold@nyctaxidataforproject.dfs.core.windows.net/trip_zone/").saveAsTable('gold.trip_zone')
```

```python
%sql
select * from gold.trip_zone
```
